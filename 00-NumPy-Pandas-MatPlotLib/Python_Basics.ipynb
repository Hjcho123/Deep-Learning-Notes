### 2.5 Lists
- A list is a built-in data type used to store an ordered, mutable (changable) collection of items. Lists allow you to keep multiple values in a single variable, accept duplicate values, and can hold different data types simultaneously (such as integers, strings, and booleans).

````python
# Create a list using square brackets
fruits = ["apple", "banana", "cherry"]

# Creating an empty list
empty_list = [] 
# Alternative: empty_list = list()

# Mixed data types
mixed_list = [42, "hello", True, 3.14]

prices = [100, 102, 98, 105, 110]

#printing a list:

print(f"Here's a list of my favorite fruits!: {fruits}")

# get the number of elements in a list:

print(len(prices))

# check if a certain element is in a list:

print(102 in prices)# is 102 a price in the price list?
````

#### 2.5.1 List Methods
- Python lists have several built-in methods for adding, removing, rearranging,
and inspecting elements.

- A **method** is a function that belongs to an object. For a list called `numbers`,
we call a method using: ```numbers.method()```

````python
# Start with a fresh list for each section

numbers = [10, 20, 30, 20, 40]

# =========================
# Adding elements
# =========================

numbers.append(50)
print(numbers)
# [10, 20, 30, 20, 40, 50]

numbers.extend([60, 70])#attaching another list to a list
print(numbers)
# [10, 20, 30, 20, 40, 50, 60, 70]

numbers.insert(1, 15) # format is (index, value)
print(numbers)
# [10, 15, 20, 30, 20, 40, 50, 60, 70]


# =========================
# Finding / counting
# =========================

print(numbers.index(20))
# 2 — index of the first 20

print(numbers.count(20))
# 2 — there are two 20s


# =========================
# Removing elements
# =========================

numbers.remove(20)
print(numbers)
# Removes the first 20

last_number = numbers.pop()
print(last_number)
# Removes and returns the last element

second_number = numbers.pop(1)
print(second_number)
# Removes and returns the element at index 1


# =========================
# Sorting
# =========================

numbers.sort()
print(numbers)
# Sorts the original list


# =========================
# Reversing
# =========================

numbers.reverse()
print(numbers)
# Reverses the original list


# =========================
# Built-in functions
# =========================

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))

sorted_numbers = sorted(numbers)
print(sorted_numbers)
````

### 2.6 Indexing and Slicing
- Python uses zero-based indexing, meaning the first item is at index 0. 
- You can also use negative indexing to look up elements starting from the end of the list (-1 represents the last item). 

````python
fruits = ["apple", "banana", "cherry"]

#INDEXING

print(fruits[0])   # Output: apple 
print(fruits[-1])  # Output: cherry (last item)
````

- You can extract a specific portion of a list by using a colon : within the brackets ```[start:stop:step]```. 
- The end index is exclusive (NOT included in the output)
- ```step``` is by default, +1

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])  # Output: [20, 30, 40]

print(numbers[:]) #prints the whole thing (not specifying a start or an end means you start from the actual beginning and end)
print(numbers[-1:-6:-1]) #reverses the list
````

- you can also slice strings the same way!

````python
word = "hello"

print(word[::-1]) # reverses the entire word
````

### 2.7 Dictionaries
- Python dictionary is a built-in data structure used to store data in key-value pairs. 
- Starting in Python 3.7, dictionaries are officially ordered (they preserve the exact order in which items are inserted) 
- they are mutable (changeable)
- they do not allow duplicate keys.

````python
# Using curly braces
stock = {
    "ticker": "AAPL",
    "price": 150,
    "sector": "Technology"
}

# Creating an empty dictionary
empty_dict = {}

print(stock["ticker"]) #gives AAPL
print(stock["price"]) #gives 150

#updating values of a key:
stock["price"] = 180
print(stock["price"]) #gives 180

#check existence of key
print("ticker" in stock) #True

#================================
#   Removing Items
#================================

# 1. del keyword: Removes the item completely
del stock["ticker"]

# 2. .pop(): Removes the key and returns its value
price = stock.pop("price") 

# 3. .popitem(): Removes and returns the last inserted key-value pair as a tuple
last_item = stock.popitem() 

# 4. .clear(): Empties the entire dictionary
stock.clear()
````

### 2.8 Tuples
- A **tuple** is an ordered collection of values, similar to a list, except that tuples are **immutable**.

````python
my_tuple = (10, 20, 30, 40)

#You can index and slice tuples exactly like lists:

print(my_tuple[0])
print(my_tuple[1:3])
````

#### Why have tuples?

- Tuples are useful when you have a collection of values that conceptually shouldn't change.

````python
#Example

stock_info = ("AAPL", 189.50, 2025)
````

#### 2.8.1 Tuple Unpacking

````python
stock = ("AAPL", 189.50, 2025)

ticker, price, year = stock

print(ticker)
print(price)
print(year)

# IMPORTANT we can assign the three variables above to set variables in the tuple in the order that they appear. i.e ticker is assigned to stock[0].
# P.S This also works with other iterables

# you can use a similar method to swap variables:

a = 10
b = 20

a, b = b, a

print(a)
print(b)
````
